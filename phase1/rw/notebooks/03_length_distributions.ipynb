<h1>Notebook Objectives</h1>

This notebook produces the assembly-length panel related to Figure 1 of the AoU-LR manuscript. It collects hifiasm report maps for haplotype-resolved assemblies and plots the distribution of hap1/hap2 N50 values across the Phase 1 cohort.

For run details, see the version of this notebook in the Researcher Workbench Controlled Tier workspace.

```python
import numpy as np
from collections import Counter
import os
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
```

```python
# Concatenate the per-sample hifiasm assembly reports (one report_map.txt per sample) into one file.
# Skipped if the file already exists; the Controlled Tier bucket is requester-pays, hence -u $GOOGLE_PROJECT.
if not os.path.exists('full_report_map.txt'):
    !gsutil -u $GOOGLE_PROJECT cat gs://fc-aou-datasets-controlled/pooled/longreads/v7_base/hifiasm/*/report_map.txt > full_report_map.txt
```

```python
# Pull the contig N50 for each haplotype: column 3 is haplotype 1, column 4 is haplotype 2.
hap1_n50 = !grep '^N50' full_report_map.txt | awk '{ print $3 }'
hap2_n50 = !grep '^N50' full_report_map.txt | awk '{ print $4 }'
```

```python
hap1_n50 = [int(a) for a in hap1_n50]
hap2_n50 = [int(a) for a in hap2_n50]
```

```python
# Histogram of haplotype-1 assembly N50 across samples (100 bp bins).
plt.figure(figsize=(10, 6))
plt.hist(hap1_n50, bins=range(int(min(hap1_n50)), int(max(hap1_n50)) + 100, 100), color='blue', alpha=0.7)
plt.title('Histogram of Genome Assembly N50 Values')
plt.xlabel('N50 (bp)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)

plt.show()
```
